# Azelaic acid / azelate: live OmniPath FastAPI extraction

This notebook queries the current OmniPath metabo FastAPI at `https://dev.omnipathdb.org/api` and exports a curated paper-supplement table set for azelaic-acid / Azelate seed entities.

It writes one CSV per curated table and one styled composite Excel workbook whose first sheet is `legend`. Seed selection is restricted to the three required records: PTFI `Azelaic acid (PTF52203)`, Recon3D `Azelaic acid; nonanedioic acid`, and the resolved `Azelate` entity with standard InChI key `BDJRBEYXGGNYIS-UHFFFAOYSA-N`.

## 1. Imports and configuration

```python
from pathlib import Path
import sys

NOTEBOOK_CWD = Path.cwd()
if (NOTEBOOK_CWD / 'azelaic_acid_query_tables.py').exists():
    sys.path.insert(0, str(NOTEBOOK_CWD))
else:
    sys.path.insert(0, str((NOTEBOOK_CWD / 'analyses' / 'azelate').resolve()))

from azelaic_acid_query_tables import (
    API_BASE,
    QUERY_TERMS,
    OUTDIR,
    run_pipeline,
)

print(API_BASE)
print(OUTDIR)
```

## 2. Query terms

The query list contains exactly the three seed handles. Search and resolve results are collapsed into the supplement `summary` sheet instead of exported as separate API-debug sheets.

```python
QUERY_TERMS
```

## 3. Live extraction and export

This step calls the FastAPI endpoints, hydrates participant entities, fetches relation evidence annotations, builds derived tables, writes CSV files, and writes the legend-first Excel workbook. Figures are generated by the rich-visualization notebook into `azelaic_acid_dev_fastapi_outputs/figures`.

```python
result = run_pipeline(
    api_base=API_BASE,
    outdir=OUTDIR,
    query_terms=QUERY_TERMS,
    progress=True,
)

tables = result.tables
print('Workbook:', result.workbook_path)
print('CSV directory:', result.csv_dir)
print('Manifest:', result.manifest_path)
print('Figures:', result.figure_paths)
```

## 5. Visualization notes

- `summary` contains all three requested seed entities and marks the PTFI/Recon3D records as `unresolved` and the InChI-key Azelate record as `resolved`.

- `relations` is the single human-readable direct-relation sheet.
- `interaction_types_by_source`, `cancer_assoc_by_sample_type`, and `fig_mindmap_data` are resolved-only figure data sheets.
