# Rich visualizations from azelaic acid / azelate FastAPI outputs

This notebook consumes the CSV tables written by `azelaic_acid_dev_fastapi_everything.ipynb` and creates a richer visualization bundle.

Figure naming is reset so the main paper outputs are `interaction_types_by_source.png` and `cancer_assoc_by_sample_type.png`. All figures are written to `azelaic_acid_dev_fastapi_outputs/figures`. Plots filter to `resolved` seed rows only; unresolved seed rows remain in the exported CSV/Excel tables. The mindmap and target-type composition are produced afterward as unlettered extra figures.

## 1. Imports and paths

```python
from pathlib import Path
import sys

NOTEBOOK_CWD = Path.cwd()
if (NOTEBOOK_CWD / 'azelaic_acid_plots.py').exists():
    sys.path.insert(0, str(NOTEBOOK_CWD))
else:
    sys.path.insert(0, str((NOTEBOOK_CWD / 'analyses' / 'azelate').resolve()))

from azelaic_acid_plots import ROOT, RICH_DIR, load_tables, make_all_visualizations

root = ROOT
rich_dir = RICH_DIR

print('Input CSV directory:', root / 'csvs')
print('Output figure directory:', rich_dir)
```

## 2. Inspect extracted table sizes

```python
tables = load_tables(root / 'csvs')
table_sizes = sorted((name, df.shape[0], df.shape[1]) for name, df in tables.items())
table_sizes[:10], len(table_sizes)
```

## 3. Generate selected rich figures

Outputs are saved in `azelaic_acid_dev_fastapi_outputs/figures` as PNG files plus `figure_index.csv` and `figure_index.json`.

```python
figure_index = make_all_visualizations(root=root, out_dir=rich_dir)
figure_index
```

## 4. Figure notes

- Main panel C: interaction-type stacked bar by source, matching the stacked-bar style in the reference image.
- Main panel D: cancer association stacked bar using unique disease/sample-type/study units; CRC contrast-row diagnostics are saved separately and the legend title is `Sample Type`.
- Extra: evidence atlas mindmap, with common-name aliases removed from ID nodes, case-only duplicate labels normalized, and unresolved seed rows filtered out.
- Extra: direct relation target-type composition by source.

```python
for _, row in figure_index.iterrows():
    label = f"Main panel {row['panel']}" if row['figure_type'] == 'main_panel' else 'Extra figure'
    print(f"{label}: {row['title']} -> {row['path']}")
```
